```python
import numpy as np
import pandas as pd
```

```python
def filter_data_frame(df : pd.DataFrame) -> pd.Series:
    df = df.loc[df.groupby("run")["round"].idxmax()]
    return df['buyer_offer']
    
```

```python
df_gemini = pd.read_csv("results/experiment1/price_trajectories.csv")
df_claude = pd.read_csv("results/experiment2/price_trajectories.csv")
df_openai = pd.read_csv("results/experiment3/price_trajectories.csv")
```

```python
gemini_price = filter_data_frame(df_gemini).reset_index(drop=True)
claude_price = filter_data_frame(df_claude).reset_index(drop=True)
openai = filter_data_frame(df_openai).reset_index(drop=True)
data = {'gemini' : gemini_price, 'claude': claude_price, 'gpt' : openai}
df = pd.DataFrame(data=data, )
df.head()
```

```python
models = df.columns
for model in models:
    print(df[model].mean())
```

```python
df.to_csv('final_price_per_models.csv')
```
